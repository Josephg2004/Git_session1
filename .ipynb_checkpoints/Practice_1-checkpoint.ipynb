# Time:
- Min reperesent%M
- PM<AM represent %p
- norml time represent %I
- railway time represent %H

# Date 
- date represent %d 
- Mondth in numbers %m
- Month in string %B (full name)
- Mont in str %b (half name)
- Year %Y(full year)
-Year %y (Half year)

```python
print(now.time())
```

```python
print(now.strftime('%M__minitus'))
print(now.strftime('%I_time'))
print(now.strftime('%H__fulltime'))
print(now.strftime('%p----AM,PM'))
print(now.strftime('%d__date'))
print(now.strftime('%A---- week name full'))
print(now.strftime('%a-----week name half'))
print(now.strftime('%m___mont in numbers'))
print(now.strftime('%B----month name in full'))
print(now.strftime('%b----month name in half'))
print(now.strftime('%y-----half year '))
print(now.strftime('%Y----Full year'))
```
